```python
def explorar_nc(path):

    ds = xr.open_dataset(path)

    print("\n========== DIMENSIONES ==========")
    for nombre, tamaño in ds.sizes.items():
        print(f"{nombre}: {tamaño}")

    print("\n========== VARIABLES ==========")
    for nombre, variable in ds.data_vars.items():
        print(
            f"{nombre}: "
            f"shape={variable.shape}, "
            f"dims={variable.dims}, "
            f"dtype={variable.dtype}"
        )

    print("\n========== COORDENADAS ==========")
    for nombre, coord in ds.coords.items():
        print(f"{nombre}: {coord.shape}")

    print("\n========== ATRIBUTOS ==========")
    print(ds.attrs)

    return ds
```

```python
ds = explorar_nc(path)
```
